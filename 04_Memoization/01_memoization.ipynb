#### Memoization in ANN

```python
# Fiboacci series

def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
fib(5)
```

```python
fib(20)
```

```python
fib(35)
```

```python
fib(40)
```

##### Memoization

```python
def fibo(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fibo(n-1, d) + fibo(n-2, d)
        return d[n]
```

```python
d = {0:1, 1:1}
fibo(40, d)
```

```python
fibo(100, d)
```
